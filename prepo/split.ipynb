```python
import os
import pandas as pd
from sklearn.model_selection import train_test_split


# Path manifest hasil cleaning
manifest_path = r"E:\Skripsi\DEC\dataset\ae-supervised-dataset\cleaned-baru\manifest.csv"

# Output folder untuk split
splits_root = r"E:\Skripsi\DEC\dataset\ae-supervised-dataset\splits-baru"
os.makedirs(splits_root, exist_ok=True)

# Load manifest
df = pd.read_csv(manifest_path)
print("Total data:", len(df))
print("Label distribusi:\n", df["label"].value_counts())
```

```python
# Train-test split (80% train+val, 20% test)
train_val_df, test_df = train_test_split(
    df, test_size=0.20, stratify=df["label"], random_state=42
)

# Dari 80% train_val, pisahkan 20% untuk validation (16% dari total)
train_df, val_df = train_test_split(
    train_val_df, test_size=0.20, stratify=train_val_df["label"], random_state=42
)

print("Train:", len(train_df))
print("Validation:", len(val_df))
print("Test:", len(test_df))

# Simpan ke CSV
train_df.to_csv(os.path.join(splits_root, "train.csv"), index=False)
val_df.to_csv(os.path.join(splits_root, "val.csv"), index=False)
test_df.to_csv(os.path.join(splits_root, "test.csv"), index=False)

print("Train/Val/Test splits saved in:", splits_root)
```
